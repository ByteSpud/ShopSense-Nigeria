```python
# ShopSense Nigeria Phone Accessories Store

# Task 1: Define your product records
products = [
    {"name":"iPhone 15 Case",    "price":2500.0,  "units_sold":42, "stock_left":3,  "is_featured":True,  "category":"Cases"},
    {"name":"Screen Protector",   "price": 800.0,  "units_sold":28, "stock_left":15, "is_featured":False, "category":"Protection"},
    {"name":"Wireless Charger",   "price":3200.0,  "units_sold":19, "stock_left":4,  "is_featured":True,  "category":"Charging"},
    {"name":"USB-C Hub",          "price":5200.0,  "units_sold": 4, "stock_left":22, "is_featured":False, "category":"Accessories"},
    {"name":"AirPod Foam Tips",   "price": 650.0,  "units_sold":35, "stock_left":8,  "is_featured":False, "category":"Audio"},
    {"name":"Pop Socket",         "price": 450.0,  "units_sold":12, "stock_left":30, "is_featured":True,  "category":"Accessories"},
    {"name":"Bluetooth Speaker",  "price":6500.0,  "units_sold": 9, "stock_left": 2,  "is_featured":False, "category":"Audio"},     # This is the 7th product that I added.
]

# What to verify
print("Type check (price should be float and units sold should be int):")       # The spaces below is for formatting purposes
print("  price:      ", type(products[0]["price"]))        # should be <class 'float'>      #The price is "2500.0" and has a decimal point and that's why it's a float
print("  units_sold: ", type(products[0]["units_sold"]))   # should be <class 'int'>
print()


# Task 2: Write classify_product()
def classify_product(product):
    """Return HOT / STEADY / SLOW"""
    sold = product["units_sold"]

    if sold >= 30:
        return "HOT"
    elif sold >= 15:
        return "STEADY"
    else:
        return "SLOW"

# Test it
print("Test:", classify_product(products[0]))  # should print HOT. # This is because "products[0]" refers to the first item in the list and 42 units were sold. Selling 30 or more units is defined as HOT.
print()


# Task 3: Loop and Print the Shop Report
separator = "═" * 48
print(separator)
print(f"{'SHOPSENSE NIGERIA — WEEKLY REPORT':^48}")
print(separator)

for p in products:                 # This "products" refers to the 7 items in my list in task 1. For loop runs through the first item (name) and when done, it goes to the second item (product). It will iterate through the 7 items in "products" until the list is empty.
    status = classify_product(p)   # This refers to the "classify_product" function in task 2, and the value will be stored in the variable for "status"
    print(f"{p['name']:<22}"
          f"₦{p['price']:>7,.0f}  "
          f"sold:{p['units_sold']:>3}  "
          f"stock:{p['stock_left']:>3}  "
          f"{status}")

print(separator)
print()

# Task 4: Generate 3 Key Business Insights
# Best seller
best_seller = products[0]                                   # Used "best_seller" instead of "best" here for the variable to help me read it better
for p in products:
    if p["units_sold"] > best_seller["units_sold"]:         # This goes through all the "units_sold" in the "products" list in task 1. This line checks if the current value here (p["units_sold"]) that it is going over in the list is greater than the current value stored in this variable (best_seller["units_sold"]). It stores the highest value in "p" and moves on to the next line. If the condition is not met, nothing happens and it goes on to the next line until the for loop is complete.
        best_seller = p

print(f"Best seller: {best_seller['name']} ({best_seller['units_sold']} sold)")     # This uses the f-string to read the output of "best_seller"

# Total stock value
total_stock_value = 0

for p in products:
    total_stock_value += p["price"] * p["stock_left"]             # This (p["units_sold"]) multiplies the price by the stock left for each product - I'll call this "V". The "sum" adds the value of "v" for every product into 1 single number.
print(f"Total stock value: ₦{total_stock_value:,.2f}")                  # This uses the f-string to read the output of "total_stock_value". I searched online for the logic of ",.0f" in task 3 and used the same logic here for ",.2f".  "," for thousands separators and ".2f" for exactly 2 decimal places 

#  Restock alert
print("Restock alert (stock_left < 5):")
for p in products:
    if p["stock_left"] < 5:                                         # For loop goes though all the "stock_left" in Task 1 product list and checks for values less than 5.
        print(f"    - {p['name']} (only {p['stock_left']} left)")   # This prints a value everytime the line above is true because of the "print()" function. The output for each time it is printed is based on the f-string code in this line
```
